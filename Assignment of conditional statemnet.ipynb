Write a program to check whether a year is a leap year.

```python
year = int(input("Enter year :"))
if year%4==0 and year%100!=0 or year%400==0:
    print("Leap year ")
else:
    print("Not leap year")
```

Write a program to check whether a character is uppercase, lowercase, digit, or special symbol.

```python
character = str(input("Enter your character : "))
if character.isupper():
    print("Uppercase")
elif character.islower():
    print("Lowercase")
elif character.isdigit():
    print("Digit")
else:
    print("Special Symbol")
    
```

Write a program to calculate grade of a student using if-elif-else.

```python
grade = int(input("Enter grade : "))

if grade>=90 and grade<=100:
    print("A grade")
elif grade>=70 and grade<=90:
    print("B grade")
elif grade>=50 and grade<=70:
    print("C grade")
elif grade>=33 and grade<=50:
    print("D grade")
else:
    print("Fail")
```
